## Task 3 : Customer Segmentation

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df = pd.read_csv(r'C:\Users\akshi\OneDrive\Desktop\Data Analysis\Data Analysis\Saiket Systems\Data\Telco_Customer_Churn_Dataset .csv')
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan).astype(float)
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df['TenureSegment'] = pd.cut(
    df['tenure'],
    bins=[0, 12, 24, 48, 72, float('inf')],
    labels=['0-1 year', '1-2 years', '2-4 years', '4-6 years', '6+ years'],
    include_lowest=True
)
```

```python
df['MonthlyChargeSegment'] = pd.cut(
    df['MonthlyCharges'],
    bins=[0, 35, 70, 90, float('inf')],
    labels=['Low', 'Medium', 'High', 'Very High'],
    include_lowest=True
)
```

```python
def analyze_segment_churn(segment_col):
    segment_churn = (
        df.groupby(segment_col, observed=False)['Churn']
        .value_counts(normalize=True)
        .unstack(fill_value=0)
    )
    segment_churn['Churn Rate'] = segment_churn['Yes'] * 100

    plt.figure(figsize=(10, 6))
    sns.barplot(x=segment_col, y='Churn Rate', data=segment_churn.reset_index())
    plt.title(f'Churn Rate by {segment_col}')
    plt.xticks(rotation=45)
    plt.ylabel('Churn Rate (%)')
    plt.xlabel(segment_col)
    plt.show()

    return segment_churn
```
